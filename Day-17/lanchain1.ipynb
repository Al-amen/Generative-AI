```python
import os
from dotenv import load_dotenv,find_dotenv
_ = load_dotenv(find_dotenv()) # read local .env file
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
```

```python
# Completion model
from langchain_groq import ChatGroq

llamaChatModel = ChatGroq(model="openai/gpt-oss-120b", api_key=GROQ_API_KEY)
```

```python
deepChatModel = ChatGroq(model="qwen/qwen3-32b", api_key=GROQ_API_KEY)
```

```python
messages = [
    ('system',"You are an historian in the Kennedy family"),
    ('human',"How many members of the family died tragically?")
]
```

```python
llamaResponse = llamaChatModel.invoke(messages)
print(llamaResponse)
```

```python
llamaResponse.content
```

```python
deepResponse = deepChatModel.invoke(messages)
print(deepResponse.content)
```

```python
# Completion model

from langchain_openai import OpenAI

llmModel = OpenAI()
```

```python
response  = llmModel.invoke("Tell me one fun fact about the Kennedy family.")
```

```python
for chunk in llmModel.stream(
    "Tell me one fun fact about the Kennedy family.i need detail response"
):
    print(chunk, end="", flush= True)
```

```python
creativeLlmModel = OpenAI(temperature=0.9)
```

```python
response  = creativeLlmModel.invoke("Tell me one fun fact about the Kennedy family.")
```

```python
print(response)
```

```python
response = llmModel.invoke(
    "Write a short 5 line poem about JFK"
)
```

```python
# Chat Model
from langchain_openai import ChatOpenAI

chatModel = ChatOpenAI(model="gpt-3.5-turbo-0125")
```

```python
messages = [
    ("system", "You are an historian expert in the Kennedy family."),
    ("human", "Tell me one curious thing about JFK."),
]
response = chatModel.invoke(messages)
```
